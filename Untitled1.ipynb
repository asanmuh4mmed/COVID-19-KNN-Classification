```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix

# ==========================================
# 1. VERİ ÖN İŞLEME (Sadece 3 Sınıf)
# ==========================================
ana_klasor = "covid_dataset/COVID-19_Radiography_Dataset"
siniflar = ['COVID', 'Normal', 'Viral Pneumonia'] # Lung Opacity çıkarıldı
IMG_SIZE = 64
MAX_RESIM_PER_CLASS = 500

X, y = [], []
print("1. Görüntüler matris formatına dönüştürülüyor...")

for label_idx, sinif in enumerate(siniflar):
    klasor_yolu = os.path.join(ana_klasor, sinif, "images")
    if not os.path.exists(klasor_yolu):
        continue
    resim_dosyalari = os.listdir(klasor_yolu)[:MAX_RESIM_PER_CLASS]

    for resim_adi in resim_dosyalari:
        resim_yolu = os.path.join(klasor_yolu, resim_adi)
        img = cv2.imread(resim_yolu, cv2.IMREAD_GRAYSCALE)
        if img is not None:
            img_resized = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
            X.append(img_resized.flatten())
            y.append(label_idx)

X = np.array(X)
y = np.array(y)
print(f"=> Okunan Veri Matrisi: {X.shape}")

# ==========================================
# 2. OPTİMİZASYON (k ve r değerlerinin bulunması)
# ==========================================
print("\n2. KNN Modeli Eğitiliyor ve Parametreler Optimize Ediliyor...")
test_oranlari = [0.30, 0.20]
k_degerleri = range(1, 16)

skorlar = {0.30: [], 0.20: []}
en_iyi_skor, opt_k, opt_r = 0, 1, 0.20

for r in test_oranlari:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=r, random_state=42)
    for k in k_degerleri:
        knn = KNeighborsClassifier(n_neighbors=k)
        knn.fit(X_train, y_train)
        skor = knn.score(X_test, y_test)
        skorlar[r].append(skor * 100)

        if skor > en_iyi_skor:
            en_iyi_skor, opt_k, opt_r = skor, k, r

print(f"=> En Başarılı Sonuç: %{int((1-opt_r)*100)} Eğitim, K={opt_k} (Doğruluk: %{en_iyi_skor*100:.2f})")

# ==========================================
# 3. GRAFİK 1: Doğruluk Karşılaştırma Grafiği
# ==========================================
plt.figure(figsize=(10, 5))
plt.plot(k_degerleri, skorlar[0.30], marker='o', color='blue', label='%70 Eğitim / %30 Test')
plt.plot(k_degerleri, skorlar[0.20], marker='s', color='darkorange', label='%80 Eğitim / %20 Test')
plt.title('3 Sınıflı Veri Seti İçin KNN Performans Analizi', fontsize=14)
plt.xlabel('K Değeri (Komşu Sayısı)', fontsize=12)
plt.ylabel('Doğruluk Oranı (%)', fontsize=12)
plt.xticks(k_degerleri)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# ==========================================
# 4. GRAFİK 2: Karmaşıklık Matrisi ve Sınıflandırma Raporu
# ==========================================
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=opt_r, random_state=42)
nihai_knn = KNeighborsClassifier(n_neighbors=opt_k)
nihai_knn.fit(X_train, y_train)
y_pred = nihai_knn.predict(X_test)

print("\n--- Hassasiyet (Precision) ve Duyarlılık (Recall) Raporu ---")
print(classification_report(y_test, y_pred, target_names=siniflar))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=siniflar, yticklabels=siniflar)
plt.title(f'Nihai KNN Karmaşıklık Matrisi (K={opt_k})', fontsize=14)
plt.xlabel('Modelin Tahmini', fontsize=12)
plt.ylabel('Gerçek Sınıf', fontsize=12)
plt.show()

# ==========================================
# 5. GRAFİK 3: Gerçek Röntgenler Üzerinde Tahmin (Görsel Şov)
# ==========================================
print("\n3. Rastgele Test Verileri Üzerinde Görsel Doğrulama:")
plt.figure(figsize=(14, 4))
rastgele_indeksler = np.random.choice(len(X_test), 4, replace=False)

for i, idx in enumerate(rastgele_indeksler):
    gercek = siniflar[y_test[idx]]
    tahmin = siniflar[y_pred[idx]]
    resim = X_test[idx].reshape(IMG_SIZE, IMG_SIZE)
    renk = 'green' if gercek == tahmin else 'red'

    plt.subplot(1, 4, i+1)
    plt.imshow(resim, cmap='gray')
    plt.title(f"Tahmin: {tahmin}\nGerçek: {gercek}", color=renk, fontweight='bold')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix

# ==========================================
# 1. VERİ ÖN İŞLEME (Sadece 3 Sınıf)
# ==========================================
ana_klasor = "covid_dataset/COVID-19_Radiography_Dataset"
siniflar = ['COVID', 'Normal', 'Viral Pneumonia'] # Lung Opacity çıkarıldı
IMG_SIZE = 64
MAX_RESIM_PER_CLASS = 500

X, y = [], []
print("1. Görüntüler matris formatına dönüştürülüyor...")

for label_idx, sinif in enumerate(siniflar):
    klasor_yolu = os.path.join(ana_klasor, sinif, "images")
    if not os.path.exists(klasor_yolu):
        continue
    resim_dosyalari = os.listdir(klasor_yolu)[:MAX_RESIM_PER_CLASS]

    for resim_adi in resim_dosyalari:
        resim_yolu = os.path.join(klasor_yolu, resim_adi)
        img = cv2.imread(resim_yolu, cv2.IMREAD_GRAYSCALE)
        if img is not None:
            img_resized = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
            X.append(img_resized.flatten())
            y.append(label_idx)

X = np.array(X)
y = np.array(y)
print(f"=> Okunan Veri Matrisi: {X.shape}\n")

# ==========================================
# 2. OPTİMİZASYON VE TABLO OLUŞTURMA
# ==========================================
print("2. KNN Modeli Eğitiliyor ve Başarı Tablosu Oluşturuluyor...\n")
test_oranlari = [0.30, 0.20]
k_degerleri = range(1, 16)

# Tablo için verileri tutacağımız sözlük
tablo_verileri = {"K Değeri": [f"K={k}" for k in k_degerleri],
                  "%70 Eğitim / %30 Test": [],
                  "%80 Eğitim / %20 Test": []}

skorlar = {0.30: [], 0.20: []}
en_iyi_skor, opt_k, opt_r = 0, 1, 0.20

for r in test_oranlari:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=r, random_state=42)
    kolon_adi = "%70 Eğitim / %30 Test" if r == 0.30 else "%80 Eğitim / %20 Test"

    for k in k_degerleri:
        knn = KNeighborsClassifier(n_neighbors=k)
        knn.fit(X_train, y_train)
        skor = knn.score(X_test, y_test)

        skorlar[r].append(skor * 100)
        tablo_verileri[kolon_adi].append(f"%{skor*100:.2f}")

        if skor > en_iyi_skor:
            en_iyi_skor, opt_k, opt_r = skor, k, r

# Pandas ile Tabloyu Ekrana Basma
df_sonuclar = pd.DataFrame(tablo_verileri)
df_sonuclar.set_index("K Değeri", inplace=True)
print("--- KNN EĞİTİM ORANI VE K DEĞERİ KARŞILAŞTIRMA TABLOSU ---")
print(df_sonuclar.to_string())
print("-" * 60)
print(f"🏆 GLOBAL OPTIMUM: %{int((1-opt_r)*100)} Eğitim, K={opt_k} (Doğruluk: %{en_iyi_skor*100:.2f})\n")

# ==========================================
# 3. GRAFİK 1: Doğruluk Karşılaştırma Grafiği
# ==========================================
plt.figure(figsize=(10, 5))
plt.plot(k_degerleri, skorlar[0.30], marker='o', color='blue', label='%70 Eğitim / %30 Test')
plt.plot(k_degerleri, skorlar[0.20], marker='s', color='darkorange', label='%80 Eğitim / %20 Test')
plt.title('3 Sınıflı Veri Seti İçin KNN Performans Analizi', fontsize=14)
plt.xlabel('K Değeri (Komşu Sayısı)', fontsize=12)
plt.ylabel('Doğruluk Oranı (%)', fontsize=12)
plt.xticks(k_degerleri)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# ==========================================
# 4. GRAFİK 2: Karmaşıklık Matrisi ve Rapor
# ==========================================
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=opt_r, random_state=42)
nihai_knn = KNeighborsClassifier(n_neighbors=opt_k)
nihai_knn.fit(X_train, y_train)
y_pred = nihai_knn.predict(X_test)

print("\n--- Hassasiyet (Precision) ve Duyarlılık (Recall) Raporu ---")
print(classification_report(y_test, y_pred, target_names=siniflar))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=siniflar, yticklabels=siniflar)
plt.title(f'Nihai KNN Karmaşıklık Matrisi (K={opt_k})', fontsize=14)
plt.xlabel('Modelin Tahmini', fontsize=12)
plt.ylabel('Gerçek Sınıf', fontsize=12)
plt.show()

# ==========================================
# 5. GRAFİK 3: Gerçek Röntgenler Üzerinde Tahmin (Görsel Şov)
# ==========================================
print("\n3. Rastgele Test Verileri Üzerinde Görsel Doğrulama:")
plt.figure(figsize=(14, 4))
rastgele_indeksler = np.random.choice(len(X_test), 4, replace=False)

for i, idx in enumerate(rastgele_indeksler):
    gercek = siniflar[y_test[idx]]
    tahmin = siniflar[y_pred[idx]]
    resim = X_test[idx].reshape(IMG_SIZE, IMG_SIZE)
    renk = 'green' if gercek == tahmin else 'red'

    plt.subplot(1, 4, i+1)
    plt.imshow(resim, cmap='gray')
    plt.title(f"Tahmin: {tahmin}\nGerçek: {gercek}", color=renk, fontweight='bold')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
